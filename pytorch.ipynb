# Tensor Attributes

Tensor attributes describe their shape, datatype, and the device on which they are stored.

```python
tensor = torch.rand(3,4)
print(tensor.shape)
print(tensor.dtype)
print(tensor.device)#device the tensor is stored on 
```

# Tensor Operations


```python
#if theres a GPU available we move the tensor to the GPU to accelerate the running
device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else 'cpu'
tensor = tensor.to(device)
print(f'device tensor is stored on: {tensor.device}')
```

## A Gentle introduction to torch.autograd

```python
import torch
from torchvision.models import resnet18, ResNet18_Weights
model = resnet18(weights=ResNet18_Weights.DEFAULT)
data = torch.rand(1,3,64,64)
labels = torch.rand(1,1000)

#resnet18 is a predefined modle and we are feeding it fake data atp
```

```python
#next run the data on the modle 
prediction = model(data)
```

```python
loss = (prediction - labels).sum()
loss.backward() # backward pass

#comparing the prediction against the fake true answer we made named labels
```

```python
#load an optimizer
optim = torch.optim.SGD(model.parameters(), lr=1e-2, momentum=0.9)
```

```python
optim.step() #gradient descent
```
