## A Gentle introduction to torch.autograd

```python
import torch
from torchvision.models import resnet18, ResNet18_Weights
model = resnet18(weights=ResNet18_Weights.DEFAULT)
data = torch.rand(1,3,64,64)
labels = torch.rand(1,1000)

#resnet18 is a predefined modle and we are feeding it fake data atp
```

```python
#next run the data on the modle 
prediction = model(data)
```

```python
loss = (prediction - labels).sum()
loss.backward() # backward pass

#comparing the prediction against the fake true answer we made named labels
```

```python
#load an optimizer
optim = torch.optim.SGD(model.parameters(), lr=1e-2, momentum=0.9)
```

```python
optim.step() #gradient descent
```

## Differentiation in Autograd

```python
#how autograd counts gradients
import torch

a = torch.tensor([2.,3.],requires_grad=True)
b = torch.tensor([6.,4.],requires_grad=True)
```

```python
#create an another tensor Q from a and b 
# Q = 3a³ - b²

Q = 3*a**3 - b**2
```

```python
external_grad = torch.tensor([1., 1.])
Q.backward(gradient=external_grad)
```

```python
# check if collected gradients are correct
print(9*a**2 == a.grad)
print(-2*b == b.grad)
```

# Neural Networks

Can be constructed using the torch.nn package
## Define the Network

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


class Net(nn.Module):

    def __init__(self):
        super().__init__()
        # 1 input image channel, 6 output channels, 5x5 square convolution
        # kernel
        self.conv1 = nn.Conv2d(1, 6, 5)
        self.conv2 = nn.Conv2d(6, 16, 5)
        # an affine operation: y = Wx + b
        self.fc1 = nn.Linear(16 * 5 * 5, 120)  # 5*5 from image dimension
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, 10)

    def forward(self, input):
        # Convolution layer C1: 1 input image channel, 6 output channels,
        # 5x5 square convolution, it uses RELU activation function, and
        # outputs a Tensor with size (N, 6, 28, 28), where N is the size of the batch
        c1 = F.relu(self.conv1(input))
        # Subsampling layer S2: 2x2 grid, purely functional,
        # this layer does not have any parameter, and outputs a (N, 6, 14, 14) Tensor
        s2 = F.max_pool2d(c1, (2, 2))
        # Convolution layer C3: 6 input channels, 16 output channels,
        # 5x5 square convolution, it uses RELU activation function, and
        # outputs a (N, 16, 10, 10) Tensor
        c3 = F.relu(self.conv2(s2))
        # Subsampling layer S4: 2x2 grid, purely functional,
        # this layer does not have any parameter, and outputs a (N, 16, 5, 5) Tensor
        s4 = F.max_pool2d(c3, 2)
        # Flatten operation: purely functional, outputs a (N, 400) Tensor
        s4 = torch.flatten(s4, 1)
        # Fully connected layer F5: (N, 400) Tensor input,
        # and outputs a (N, 120) Tensor, it uses RELU activation function
        f5 = F.relu(self.fc1(s4))
        # Fully connected layer F6: (N, 120) Tensor input,
        # and outputs a (N, 84) Tensor, it uses RELU activation function
        f6 = F.relu(self.fc2(f5))
        # Fully connected layer OUTPUT: (N, 84) Tensor input, and
        # outputs a (N, 10) Tensor
        output = self.fc3(f6)
        return output


net = Net()
print(net)
```

```python
#learnable parameters are returned by net.parameters()

params = list(net.parameters())
print(len(params))
print(params[0].size())
```

```python
input = torch.randn(1,1,32,32)
out = net(input)
print(out)
```

```python
#Zero the gradient buffers of all parameters and backprops with random gradients

net.zero_grad()
out.backward(torch.randn(1,10))
```

## Loss Function

```python
#this guesses how wrong the networds guess is 
output = net(input) #actually runs the image throuhgh the network now 
target = torch.randn(10) #fake correct lable for the netwrok to practice like compare to  
target = target.view(1,-1) #make it the same shape as the output
criterion = nn.MSELoss() #sets up the a way to measure error 

loss = criterion(output, target)
print(loss)
```

```python
print(loss.grad_fn)  # MSELoss
print(loss.grad_fn.next_functions[0][0])  # Linear
print(loss.grad_fn.next_functions[0][0].next_functions[0][0])  # ReLU
```

## Backprop

```python
net.zero_grad()     # zeroes the gradient buffers of all parameters

print('conv1.bias.grad before backward')
print(net.conv1.bias.grad)

loss.backward()

print('conv1.bias.grad after backward')
print(net.conv1.bias.grad)
```

# Training a Classifier
## Training an image classifier

```python
#load and normalise CIFAR10 - training and test datasets
import torch
import torchvision 
from torchvision.transforms import v2
```

```python
#getting CIFAR10 images ready to feed into the network
transform = v2.Compose([
    v2.ToImage(),
    v2.ToDtype(torch.float32, scale=True),
    v2.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))])

batch_size = 4

trainset = torchvision.datasets.CIFAR10(root='./data', train=True,
                                        download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=batch_size,
                                          shuffle=True, num_workers=2)

testset = torchvision.datasets.CIFAR10(root='./data', train=False,
                                       download=True, transform=transform)
testloader = torch.utils.data.DataLoader(testset, batch_size=batch_size,
                                         shuffle=False, num_workers=2)

classes = ('plane', 'car', 'bird', 'cat',
           'deer', 'dog', 'frog', 'horse', 'ship', 'truck')

        
```

This part sets up the data pipeline — getting CIFAR-10 images ready to feed into a network.

transform — a mini pipeline applied to every image:

ToImage() — converts it into a proper image tensor
ToDtype(torch.float32, scale=True) — turns pixel values (0–255) into floats scaled to 0–1
Normalize((0.5,0.5,0.5), (0.5,0.5,0.5)) — shifts those values so they're centered around 0 instead of 0–1 (helps training go smoother)

trainset / testset — downloads the CIFAR-10 dataset (60,000 tiny 32×32 color photos across 10 categories), split into a training portion and a testing portion, applying the transform above to each image as it's loaded.

trainloader / testloader — wraps each dataset so you can pull it in small batches (batch_size=4 images at a time) instead of one image at a time. shuffle=True on the training set mixes up the order each epoch so the network doesn't learn any pattern from image ordering; the test set doesn't need shuffling since you're just evaluating, not learning. num_workers=2 loads batches in parallel in the background so the CPU isn't a bottleneck while the GPU trains.

classes — just a lookup table: the dataset gives you numeric labels (0–9), and this tuple lets you turn label 2 back into "bird" for display purposes.

```python
import matplotlib.pyplot as plt
import numpy as np

# functions to show an image


def imshow(img):
    img = img / 2 + 0.5     # unnormalize
    npimg = img.numpy()
    plt.imshow(np.transpose(npimg, (1, 2, 0)))
    plt.show()


# get some random training images
dataiter = iter(trainloader)
images, labels = next(dataiter)

# show images
imshow(torchvision.utils.make_grid(images))
# print labels
print(' '.join(f'{classes[labels[j]]:5s}' for j in range(batch_size)))
```
