**Tensor Initialization**

 With 60 mins blitz from official pytorch site

```python
#can be initialised in different ways _ this here is direclty from data
data = [[1,2],[3,4]]
x_data = torch.tensor(data)
x_data
```

```python
import numpy as np
#From a numpy array
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
x_np
```

```python
#from another tensor  
x_ones = torch.ones_like(x_data) #retains the properties of the initial tensor
print(f'Ones tensor :{x_ones}')
```

```python
x_rand = torch.rand_like(x_data,dtype=torch.float)
print(f'Random tensor {x_rand}')

#overrides the datatype of x_data but keeps the same dimensions 
#the data differs - now float
```

```python
shape = (2,3,)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)
```
